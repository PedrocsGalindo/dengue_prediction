```python
import pandas as pd
from pathlib import Path 
```

```python
current_dir = Path(Path.cwd()).parent.parent
data_dir = current_dir / "data"
source_data_dir = data_dir / "source"
print(data_dir)
```

```python

import re
import unicodedata
 
def show_columns(df, n_range = 7):
    df = list(df.columns)
    l = []
    range = n_range
    for c in df:
        if range == n_range:
            print(l)
            l = []
            range = 0
        range += 1
        l.append(c)
    print("\n")

def normalize_column_name(col):
    # tira acentos
    col = unicodedata.normalize("NFKD", col).encode("ascii", "ignore").decode("utf-8")
    # minúsculas
    col = col.lower()
    # troca qualquer coisa que não seja letra ou número por _
    col = re.sub(r"[^a-z0-9]+", "_", col)
    # remove _ no começo e no fim
    col = col.strip("_")
    return col
```

```python
dengue_dir = source_data_dir / "DENGBR"
df_23 = pd.read_csv(dengue_dir / "DENGBR23.csv", sep=",")
```

```python
# remove coluns with more than 40% missing values
df_23 = df_23.loc[
    :, df_23.isnull().mean() < 0.4
]
df_23.columns = [normalize_column_name(col) for col in df_23.columns]

show_columns(df=df_23)
print(f"porcentagem de valores ausentes:\n{(df_23.isnull().mean() * 100).sort_values(ascending=False)}%")
print(df_23.info())
df_23.head()
```
