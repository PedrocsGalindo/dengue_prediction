````python
import pandas as pd
import numpy as np

from pathlib import Path 
````

````python
current_dir = Path(Path.cwd()).parent
data_dir = current_dir / "data"
dataset_dir = data_dir / "processed/datasets"
````

````python
from sklearn.model_selection import TimeSeriesSplit
ds = pd.read_csv(dataset_dir / "recife_dataset.csv")

X = ds.drop(columns=['casos_dengue'])
y = ds['casos_dengue']

tscv = TimeSeriesSplit(n_splits=5)
for train_idx, test_idx in tscv.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
````

## TPOT
    evolucao genética

````python
from tpot import TPOTRegressor

tpot = TPOTRegressor(
    generations=5,
    population_size=20,
    verbosity=2
)

tpot.fit(X_train, y_train)
tpot.score(X_test, y_test)
````

## H2O AutoML 
    produção

````python
import h2o
from h2o.automl import H2OAutoML

h2o.init()

data = h2o.import_file("dados.csv")

train, test = data.split_frame(ratios=[0.8])

x = data.columns[:-1]
y = data.columns[-1]

aml = H2OAutoML(max_models=10, seed=1)

aml.train(x=x, y=y, training_frame=train)

leader = aml.leader
print(leader.model_performance(test))
````
